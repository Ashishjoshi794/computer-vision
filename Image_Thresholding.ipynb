```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("/content/drive/MyDrive/my pic.jpeg")
plt.imshow(img)
```

```python
img = cv2.imread("/content/drive/MyDrive/my pic.jpeg",0)
plt.imshow(img,cmap='gray')
```

threshold.jpeg

```python
lim,img1 = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY) # 127 is the threshold value and 255 is the maxmim value
plt.imshow(img1, cmap="gray")
```

```python
lim,img1 = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY_INV) # 127 is the threshold value and 255 is the maxmim value
plt.imshow(img1, cmap="gray")
```

```python
lim,img1 = cv2.threshold(img, 127, 255, cv2.THRESH_TRUNC) # 127 is the threshold value and 255 is the maxmim value
plt.imshow(img1, cmap="gray")
```

```python
lim,img1 = cv2.threshold(img, 127, 255, cv2.THRESH_TOZERO) # 127 is the threshold value and 255 is the maxmim value
plt.imshow(img1, cmap="gray")
```

```python
lim,img1 = cv2.threshold(img, 127, 255, cv2.THRESH_TOZERO_INV) # 127 is the threshold value and 255 is the maxmim value
plt.imshow(img1, cmap="gray")
```
