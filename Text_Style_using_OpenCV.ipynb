```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = np.zeros(shape=(500,500), dtype = np.int32)
plt.imshow(img, cmap='gray')
```

```python
cv2.putText(img,"A",(100,400),cv2.FONT_HERSHEY_PLAIN,fontScale=30,color=(255,255,255),thickness=30)
plt.imshow(img, cmap='gray')
```

```python
cv2.putText(img,"JOSHI",(10,450),cv2.FONT_HERSHEY_PLAIN,fontScale=3,color=(255,255,255),thickness=13)
plt.imshow(img, cmap='gray')
```

```python
img1 = np.zeros(shape=(500,500), dtype = np.int32)
plt.imshow(img1, cmap='gray')
```

```python
cv2.putText(img1,"A",(90,450),cv2.FONT_HERSHEY_PLAIN,fontScale=30,color=(255,255,255),thickness=30)
cv2.putText(img1,"A",(90,450),cv2.FONT_HERSHEY_PLAIN,fontScale=30,color=(0,0,0),thickness=13)
plt.imshow(img1, cmap='gray')
```

```python
cv2.putText(img1,"ASHISH JOSHI",(10,450),cv2.FONT_HERSHEY_PLAIN,fontScale=3,color=(255,255,255),thickness=13)

cv2.putText(img1,"ASHISH JOSHI",(10,450),cv2.FONT_HERSHEY_PLAIN,fontScale=3,color=(0,0,0),thickness=3)
plt.imshow(img1, cmap='gray')
```
