```python
import cv2
import matplotlib.pyplot as plt
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
img = cv2.imread("/content/drive/MyDrive/my pic.jpeg")
```

```python
plt.imshow(img)
```

```python
pic = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(pic)
```

```python
pic1 = cv2.cvtColor(pic, cv2.COLOR_RGB2HSV)  # hue saturation value
plt.imshow(pic1)
```

```python
pic2 = cv2.cvtColor(pic, cv2.COLOR_RGB2HLS) #hue saturation lightness
plt.imshow(pic2)
```

```python
pic2 = cv2.cvtColor(pic, cv2.COLOR_BGR2HLS) #hue saturation lightness
plt.imshow(pic2)
```

```python
pic2 = cv2.cvtColor(pic, cv2.COLOR_) #hue saturation lightness
plt.imshow(pic2)
```
