```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
image1 = cv2.imread("/20star.jpeg")
plt.imshow(image1)
```

```python
image2 = cv2.imread("/5star.jpeg")
plt.imshow(image2)
```

```python
height, width, _ = image1.shape
resized_image2 = cv2.resize(image2, (width, height))
add_img = cv2.add(image1, resized_image2)
plt.imshow(add_img)
```

```python
weightedimg = cv2.addWeighted(image1, 0.5, resized_image2, 0.5, 0)
plt.imshow(weightedimg)
```

```python
weightedimg = cv2.addWeighted(image1, 1, resized_image2, 0.5, 0)
plt.imshow(weightedimg)
```

```python
weightedimg = cv2.addWeighted(image1, 1, resized_image2, 0.5, 100) # gama value apply hole image (eg:there is a last value 100)
plt.imshow(weightedimg)
```

```python
subimg = cv2.subtract(image1, resized_image2)
plt.imshow(subimg)
```

```python
mutimg = cv2.multiply(image1, resized_image2)
plt.imshow(mutimg)
```

```python
andimg = cv2.bitwise_and(image1, resized_image2)
plt.imshow(andimg)
```

```python
image4 = cv2.imread("/20 star bihaind.jpeg")
plt.imshow(image4)
```

```python
height, width, _ = image1.shape # dutai image ko coman white part show hunxa
resized_image4 = cv2.resize(image4, (width, height))
andimg = cv2.bitwise_and( resized_image2, resized_image4)
plt.imshow(andimg)
```

```python
orimg = cv2.bitwise_or(image1, resized_image4) # or operation yo la purai part show garxa
plt.imshow(orimg)
```

```python
notimg = cv2.bitwise_not(image1) # white to black and black to white it ia a not operation
plt.imshow(notimg)
```

```python
notimg = cv2.bitwise_not(resized_image4)
plt.imshow(notimg)
```

```python
xorimg = cv2.bitwise_xor( resized_image2, resized_image4) # yo la comon area lai black hunxa
plt.imshow(xorimg)
```
