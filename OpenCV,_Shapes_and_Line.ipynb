```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img1 = np.zeros(shape=(500,500,3), dtype = np.int32)
```

```python
img1.shape
```

```python
plt.imshow(img1)
```

```python
cv2.rectangle(img1,pt1=(300,50),pt2=(400,250),color=(0,0,255),thickness=5)
plt.imshow(img1)
```

```python
cv2.rectangle(img1,pt1=(200,30),pt2=(500,350),color=(0,255,0),thickness=5)
plt.imshow(img1)
```

```python
cv2.circle(img1, center=(30,50), radius=25, color=(220,0,0),thickness=5)
plt.imshow(img1)
```

```python
cv2.circle(img1, center=(50,70), radius=25, color=(220,0,0),thickness=-1)
plt.imshow(img1)
```

```python
cv2.line(img1,pt1=(500,500),pt2=(0,0),color=(255,255,255),thickness=5)
plt.imshow(img1)
```

```python
img2 = np.zeros(shape=(500,500,3), dtype = np.int32)
plt.imshow(img2)
```

```python
font = cv2.FONT_HERSHEY_PLAIN
cv2.putText(img2, text='ASHISH joshi',org=(10,500),fontFace=font,fontScale=3,color=(255,255,255),thickness=3,lineType=cv2.LINE_AA)
plt.imshow(img2)
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(img2, text='ASHISH joshi',org=(50,200),fontFace=font,fontScale=3,color=(255,255,255),thickness=3,lineType=cv2.LINE_AA)
plt.imshow(img2)
```

```python
img3 = np.zeros(shape=(500,500,3), dtype = np.int32)
plt.imshow(img3)
```

```python
pts = np.array([[100,300],[200,200],[400,300],[200,400]], np.int32)
cv2.polylines(img3,[pts],isClosed=True,color=(0,255,255),thickness=5)
plt.imshow(img3)
```

```python
img4 = np.zeros(shape=(500,500,3), dtype = np.int32)
plt.imshow(img4)
```

```python
pts1 = np.array([[100,300],[200,200],[400,300],], np.int32)
cv2.polylines(img4,[pts1],isClosed=True,color=(250,255,255),thickness=5)
plt.imshow(img4)
```

```python
pts1 = np.array([[100,300],[200,200],[400,300],[500,300],[500,100]], np.int32)
cv2.polylines(img4,[pts1],isClosed=True,color=(250,255,255),thickness=5)
plt.imshow(img4)
```
