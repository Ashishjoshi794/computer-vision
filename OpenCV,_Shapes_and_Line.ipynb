```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Black image
img3 = np.zeros((300,300,3), dtype=np.uint8)

# Function to draw a small star
def draw_star(img, x, y, s=8):
    pts = np.array([
        [x, y-s],
        [x+s//3, y-s//3],
        [x+s, y-s//3],
        [x+s//2, y+s//4],
        [x+2, y+s],
        [x, y+s//2],
        [x-2, y+s],
        [x-s//2, y+s//4],
        [x-s, y-s//3],
        [x-s//3, y-s//3]
    ], np.int32)

    cv2.polylines(img, [pts], True, (255,255,255), 1)

# Draw 30 small stars
np.random.seed(42)  # same result every time
for _ in range(30):
    x = np.random.randint(12, 288)
    y = np.random.randint(12, 288)
    draw_star(img3, x, y, s=8)

# Display
plt.imshow(cv2.cvtColor(img3, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Black image
img = np.zeros((400, 400, 3), dtype=np.uint8)

# 5-point star function
def draw_star(img, x, y, size=10):
    pts = np.array([
        [x, y-size],
        [x+3, y-3],
        [x+size, y-3],
        [x+5, y+3],
        [x+7, y+size],
        [x, y+6],
        [x-7, y+size],
        [x-5, y+3],
        [x-size, y-3],
        [x-3, y-3]
    ], np.int32)

    cv2.polylines(img, [pts], True, (255, 255, 255), 1)

# Draw 20 stars in a 5×4 grid
for row in range(4):
    for col in range(5):
        draw_star(img, 40 + col*70, 50 + row*80, size=10)

# Show image
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Blank image
img3 = np.zeros((300,300,3), dtype=np.uint8)

# Original star points
points = np.array([
    [150, 40],
    [170,100],
    [240,100],
    [185,145],
    [205,220],
    [150,175],
    [95,220],
    [115,145],
    [60,100],
    [130,100]
], np.int32)

# Center and scale the star
center = np.array([150,130])
star = ((points - center) * 0.2).astype(np.int32)

# Draw 20 stars (5 columns × 4 rows)
for row in range(4):
    for col in range(5):
        offset = np.array([35 + col*55, 35 + row*65])
        cv2.polylines(img3, [star + offset], True, (255,255,255), 1)

plt.imshow(cv2.cvtColor(img3, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Blank image
img3 = np.zeros((400, 400, 3), dtype=np.uint8)

# 20-point star
center = (200, 200)
outer_r = 140
inner_r = 60

points = []
for i in range(40):  # 20 outer + 20 inner vertices
    angle = np.pi/2 + i * np.pi/20
    r = outer_r if i % 2 == 0 else inner_r
    x = int(center[0] + r * np.cos(angle))
    y = int(center[1] - r * np.sin(angle))
    points.append([x, y])

points = np.array(points, np.int32)

# Fill the star with white
cv2.fillPoly(img3, [points], (255, 255, 255))

plt.imshow(cv2.cvtColor(img3, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
